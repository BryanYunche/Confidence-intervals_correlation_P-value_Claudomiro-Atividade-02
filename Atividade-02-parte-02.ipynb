## Dataset sobre performace escolar

```python
# Recomendo baixar a biblioteca de datasets do UCI
# pip install ucimlrepo
# https://archive.ics.uci.edu/dataset/320/student+performance

from ucimlrepo import fetch_ucirepo 
import pandas as pd
import numpy as np
  
# fetch dataset 
student_performance = fetch_ucirepo(id=320) 
  
# data (as pandas dataframes) 
x = student_performance.data.features
y = student_performance.data.targets 

# metadata 
# print(student_performance.metadata) 
  
# variable information 
print(student_performance.variables) 
```

### Selecionando a feature que irei utilizar como amostra

```python
# ==============================================================================
# Agora vou transformar em um dataframe a minha amostra
df = student_performance.data.features

# Irei utilizar o tempo de estudo por semana como amostra
tempo_estudo = df['studytime']

# A respeito dessa feature
# studytime - weekly study time (numeric: 1 - <2 hours, 2 - 2 to 5 hours, 3 - 5 to 10 hours, or 4 - >10 hours)

# print(tempo_estudo)
```

### Calculando a média e o desvio padrão da amostra

```python
media = np.mean(tempo_estudo)
desvio_padrao = np.std(tempo_estudo)

print(media)
print(desvio_padrao)
```
