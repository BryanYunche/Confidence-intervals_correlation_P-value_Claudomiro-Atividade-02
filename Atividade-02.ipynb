## Função "linregress"
### Utilidade da função:
- Basicamente, a função é utilizada para traçar uma reta através de um conjunto de pontos.
- Em termos práticos ela utiliza a equação da reta, _y = ax + b_, pra desenhar essa reta.
- Mas qualquer um pode traçar uma reta através de pontos, então ela faz mais que isso, ela minimiza a distância real entre a reta e os pontos

### Como ela funciona para desenhar a reta
- Ela tem como entrada dois conjuntos de pontos X e Y
- E retorna cinco valores, mas os mais importante são o "slope" e o "intercept" que são respectivamente o _a_ e o _b_ da equação da reta, _y = ax + b_.
- Quando conseguirmos os valore do "slope" e o "intercept", o numpy vai nos permitir fazer o calculo de todos os pontos da reta para o xi dado.
- Note que não tenho a reta imediatamente, mas apenas os meios para calcula-la!

### Outras saídas que a função linregress nos dá fora o da equação da reta
- "rvalue" que é o Coeficiente de Correlação que mede a força da relação entre os valores de xi e yi que vai de 1 a 0
- "pvalue" é o P-Value mede a se é real ou fruto do acaso a correlação entre os dados. ~~Pelo o que eu entendi~~.
- "stderr" é o erro padrão dos dados apresentados, basicamente ele mede a incerteza da inclinação da reta, "slope", ou seja, se esse valor for muito alto, um dado novo pode mudar de forma drastica a reta!

# Plotando Gráficos com Linregress!

```python
# Gráfico 01

# Usando a Função linregress
res_1 = stats.linregress(x1, y1)

# Análisando
print(f"Coeficiente de Correlação: {res_1.rvalue:8f}")
print(f"Erro Padrão: {res_1.stderr:8f}")


# Fórmula da equação geral da y = ax + b
reta_1 = res_1.slope * x1 + res_1.intercept

# Plot do gráico
plt.scatter(x1, y1, color='blue')

# Plotando a reta da regressão
plt.plot(x1, reta_1, color='red', linewidth=2, label=f'Regressão (p={res_1.pvalue:.3f})')


plt.title("Gráfico 01")
plt.legend()
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
```

```python
# Gráfico 02

# Usando a Função linregress com stats garante que eu estou pegando os valores corretamente
res_2 = stats.linregress(x2, y2)

# Análisando o P-value
print(f"Coeficiente de Correlação: {res_2.rvalue:8f}")
print(f"Erro Padrão: {res_2.stderr:8f}")

# Fórmula da equação geral da y = ax + b
reta_2 = res_2.slope * x2 + res_2.intercept

# Plot do gráico
plt.scatter(x2, y2, color='red')

# Plotando a reta da regressão
plt.plot(x2, reta_2, color='blue', linewidth=2, label=f'Regressão (p={res_2.pvalue:.3f})')


plt.title("Gráfico 02")
plt.legend()
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
```

## Parte 02 - Estimando um Intervalo de Confiança para a Média
- Obtenha um conjunto de dados amostrais (por exemplo, pesos de indivíduos, notas de testes, etc.).
- Calcule o intervalo de confiança de 95% para a média da população seguindo os passos:
- Calcular a média e o desvio padrão da amostra.
- Usar a fórmula para intervalo de confiança ou a biblioteca scipy.stats.

Desafio: Repetir o processo para diferentes níveis de confiança (por exemplo, 90%, 99%) e comparar os resultados. Visualizar os intervalos de confiança em um gráfico usando matplotlib para mostrar a variabilidade.

### Dataset sobre Risco de Saúde para Mulheres Grávidas

```python
# https://archive.ics.uci.edu/dataset/863/maternal+health+risk

!pip install ucimlrepo

from ucimlrepo import fetch_ucirepo
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st

# fetch dataset
maternal_health_risk = fetch_ucirepo(id=863)

# data (as pandas dataframes)
X = maternal_health_risk.data.features
y = maternal_health_risk.data.targets

# metadata
print(maternal_health_risk.metadata)

# variable information
print(maternal_health_risk.variables)
```

### Selecionando a feature como amostra e calculando a Média e o Desvio Padrão

```python
# Agora vou transformar em um dataframe a minha amostra
df = maternal_health_risk.data.features

# Irei utilizar o tempo de estudo por semana como amostra
idade = df['Age']

#Calculando a Média e o Desvio Padrão
media = np.mean(idade)
desvio_padrao = np.std(idade)
tamanho_amostra = len(idade)

print(media)
print(desvio_padrao)
print(tamanho_amostra)
```

### Analizando o histograma da amostra

```python
sns.histplot(data=idade, bins=20, color='blue', alpha=0.5, kde=True)

# Legendas do Histograma
plt.title('Distribuição de mulheres Grávidas por Idade')
plt.xlabel('Idade')
plt.ylabel('Frequência')

# Show plot
plt.show()
```

### Consultando o valor de Z para os niveis de confiança 95%, 90% e 99%

```python
alfa_95 = 0.05 # É a minha probabilidade de erro, isso é, para uma confiança de 95% a minha probabilida de erro terá de ser 5%
alfa_90 = 0.1 # É a minha probabilidade de erro, isso é, para uma confiança de 90% a minha probabilida de erro terá de ser 10%
alfa_99 = 0.01 # É a minha probabilidade de erro, isso é, para uma confiança de 99% a minha probabilida de erro terá de ser 1%

z_95 = st.norm.ppf(1 - alfa_95/2)
z_90 = st.norm.ppf(1 - alfa_90/2)
z_99 = st.norm.ppf(1 - alfa_99/2)

#print(z_95)
#print(z_90)
#print(z_99)
```

### Aplicando a formula Para calcular os Intervalos de Confiança

```python
intervalo_confianca_superior_95 = media + z_95 * (desvio_padrao / np.sqrt(tamanho_amostra))
intervalo_confianca_inferior_95 = media - z_95 * (desvio_padrao / np.sqrt(tamanho_amostra))


print("INTERVALO DE CONFIANÇA PARA 95% DE ACERTO")
print("Limite Superior: ", intervalo_confianca_superior_95.round(3))
print("Média: ", media.round(3))
print("Limite Inferior: ", intervalo_confianca_inferior_95.round(3))

################################################################################

intervalo_confianca_superior_90 = media + z_90 * (desvio_padrao / np.sqrt(tamanho_amostra))
intervalo_confianca_inferior_90 = media - z_90 * (desvio_padrao / np.sqrt(tamanho_amostra))


print("INTERVALO DE CONFIANÇA PARA 90% DE ACERTO")
print("Limite Superior: ", intervalo_confianca_superior_90.round(3))
print("Média: ", media.round(3))
print("Limite Inferior: ", intervalo_confianca_inferior_90.round(3))

################################################################################

intervalo_confianca_superior_99 = media + z_99 * (desvio_padrao / np.sqrt(tamanho_amostra))
intervalo_confianca_inferior_99 = media - z_99 * (desvio_padrao / np.sqrt(tamanho_amostra))


print("INTERVALO DE CONFIANÇA PARA 99% DE ACERTO")
print("Limite Superior: ", intervalo_confianca_superior_99.round(3))
print("Média: ", media.round(3))
print("Limite Inferior: ", intervalo_confianca_inferior_99.round(3))
```

### Visualização da variabilidade do intervalo de confiança

```python
# Histograma com Intervalo de Confiança de 95%
sns.histplot(data=idade, bins=20, color='blue', alpha=0.5, kde=True)

# Legendas do Histograma
plt.title('Intervalo de Confiança de 95% para a média de idade de mulheres grávidas')
plt.xlabel('Idade')
plt.ylabel('Frequência')

# Criando uma sombra para definir o Intervalo de Confiança
plt.axvspan(intervalo_confianca_inferior_95, intervalo_confianca_superior_95, color='green', alpha=0.3)

# Linha para a média central
plt.axvline(media, color='red', linestyle='--', label=f'Média: {media:.2f}')

# Show plot
plt.show()
```

```python
# Histograma com Intervalo de Confiança de 90%
sns.histplot(data=idade, bins=20, color='blue', alpha=0.5, kde=True)

# Legendas do Histograma
plt.title('Intervalo de Confiança de 90% para a média de idade de mulheres grávidas')
plt.xlabel('Idade')
plt.ylabel('Frequência')

# Criando uma sombra para definir o Intervalo de Confiança
plt.axvspan(intervalo_confianca_inferior_90, intervalo_confianca_superior_90, color='green', alpha=0.3)

# Linha para a média central
plt.axvline(media, color='red', linestyle='--', label=f'Média: {media:.2f}')

# Show plot
plt.show()
```

```python
# Histograma com Intervalo de Confiança de 99%
sns.histplot(data=idade, bins=20, color='blue', alpha=0.5, kde=True)

# Legendas do Histograma
plt.title('Intervalo de Confiança de 99% para a média de idade de mulheres grávidas')
plt.xlabel('Idade')
plt.ylabel('Frequência')

# Criando uma sombra para definir o Intervalo de Confiança
plt.axvspan(intervalo_confianca_inferior_99, intervalo_confianca_superior_99, color='green', alpha=0.3)

# Linha para a média central
plt.axvline(media, color='red', linestyle='--', label=f'Média: {media:.2f}')

# Show plot
plt.show()
```

## Prática 03 - P-value em um Teste de Hipótese

Forneça dois conjuntos de dados: um para um grupo de controle e outro para um grupo de teste.
- Formule a hipótese nula.
- Realizar um teste t de duas amostras usando scipy.stats.ttest_ind.

Interpretar o p-value resultante e decidir se rejeitam a hipótese nula em um nível de significância dado (por exemplo, 0,05).




### Considerações sobre a questão

Assisti a aula gravada: https://drive.google.com/file/d/1af490hg8sZAsPdCRhNGGqYTTx--uPSxs/view

Como o processo mais simples para realizar um t teste é utilizando números tipo quantitativos como idade, peso, etc.. eu irei seguir com os dados de idade das mulheres grávidas.

A base de dados me oferece um valor categórico de risco alto e risco baixo para a saúde de mulheres grávidas. então irei separa em dois grupos.

- Amostra 01: Grávidas com baixo risco de saúde
- Amostra 02: Grávidas com alto risco de saúde

**A Hipotese Núla** que quero provar é ***"A idade média das grávidas de alto risco é igual à idade média das grávidas de baixo risco."***



```python
# Mais cedo no nosso código separamos os resultados do datfram em y e as features em x, seguindo o padrão de uso da UCI
# Mas para esse código vamos precisar dos resultados para criar os dois grupos de amostra, para isso vamos unificar novamente o data frame
df = X.copy()
df['RiskLevel'] = y['RiskLevel'].values

#Agora que pegamos os valores de Níveis de Risco, vamos filtrar e obter as duas amostras.
amostra_controle = df[df['RiskLevel'] == 'low risk']['Age']
amostra_teste = df[df['RiskLevel'] == 'high risk']['Age']

# Agora vamos pegar a média de idade da população
idade_populacao = df['Age']
media_idade_populacao = np.mean(idade_populacao)

# A própria base de dados informa que não existem valores null, vamos confiar então kkkkk
# Mas vamos verificar o tamanho de cada grupo
print("Tamanho da amostra de controle: ", len(amostra_controle))
print("Tamanho da amostra de teste: ", len(amostra_teste))
print("Tamanho da população: ", len(df['Age']))
print("Média da população: ", media_idade_populacao.round(3))

#Precisamos das médias de idade das amostras
print("Média da amostra de controle: ", np.mean(amostra_controle).round(3))
print("Média da amostra de teste: ", np.mean(amostra_teste).round(3))


# Agora só aplicar o famoso t test
# Perceba que eu não preciso calcular as médias das amostras, a função já faz isso!
t_statistic, p_value = st.ttest_ind(amostra_controle, amostra_teste)

print("Valor t: ", t_statistic)
print("P-value: ", p_value)

#Defino aqui o valor de confiança que eu quero para a minha amostra em porcentagem
confianca = 95

# Dado o valor de confiança vou calcular meu nível de significancia
alpha = 1 - (confianca/100)

print("Hipotese Nula: A idade média das grávidas de alto risco é igual à idade média das grávidas de baixo risco.")
if p_value < alpha:
    print("Rejeitar a hipótese nula")
else:
    print("Não rejeitar a hipótese nula")
```

## ANÁLISE DOS RESULTADOS DA PRÁTICA 03

### Interpretando o código, vou utilizar comentátios pra fazer a análise pois não consegui trabalhar legal a explicação em markdown.

```python
# Começando pelas bibliotecas
# Precisamos da biblioteca do Numpy para fazer calculos com baixo custo operacional, haja visto que roda C por baixo disso.
import numpy as np

# Importamos a bibliotexa do scipy.stats e importamos em especifico o ttest_ind para fazer a análise das amostras
from scipy.stats import ttest_ind

# Aqui está definindo quantos testes serão feitos para identificar a proporção
num_simulations = 10000

# Aqui está sendo definido qual vai ser o tamanho da amostra
sample_size = 30

# Aqui está sendo definido o valor REAL da média
true_mean = 50

# Aqui eu acredito que o valor real do desvio padrão
true_std = 10

# O nível de siguinificancia é baiscamente aquele valor de corte do P-value que determina quanto de erro é permitido para ser considerado uma valor estatisticamente relevante.
significance_level = 0.05

# Isso aqui é basicamente o contador de casos, pois os resultados apesar de serem númericos, eles estão sendo análisados de forma categórica agora, como sim ou não.
significant_results = 0


for _ in range(num_simulations):

    # Aqui está apenas criando os dois grupos de amostra de forma randomica.
    # É importante notar que apesar de estar sendo feito de forma randomica ele está normalizando esses dados, passando o desvio padrão e uma média, ou seja, mesmo os números aleatórios terão a distribuição normal parecida.
    group1 = np.random.normal(true_mean, true_std, sample_size)
    group2 = np.random.normal(true_mean, true_std, sample_size)

    # Aqui vai está a função que vai comparar se os dois grupos de amostra gerados apresentam uma relação estatistica relavante
    t_stat, p_value = ttest_ind(group1, group2)

    # Aqui que está sendo feito a "conversão" númerico para categórico, como explicado no exemplo feito em aula com os batimentos cardiacos das mulheres.
    if p_value < significance_level:
        significant_results += 1

# Aqui é o calculo da proporção binomial de acordo com a formula alternativa ao t teste esplicado em aula no exemplo dos batimentos cardiacos explicado.
# Basicamente espera-se encontrar um valor próximo ao nível de significancia definido, isso caso os dois grupos sejam iguais.
proportion_significant = significant_results / num_simulations

# Quantidade de grupos testados entre si
print(f"Number of simulations: {num_simulations}")

# Nível de confiança estabelecido
print(f"Significance level (alpha): {significance_level}")

# Não entendi o que seria o erro tipo I, mas acredito que seja os 5% de erro descoberto na proporção
print(f"Proportion of significant results (Type I error rate): {proportion_significant:.4f}")

```

### Análise dos resultados para 5% de siguinificância!
- Como os grupos foram criados com o mesmo desvio padrão, média e tamanho, mesmo que sejam números aleatórios podemos considerar que são números da mesma população.
- Outra coisa interssante de se perceber é que como são amostras da mesma população, quando fazemos um número exprecivo de comparação entre as amostras elas tendem a se comportar com o mesmo nível de siguinificancia dado, isto é, podem haver amostras que não são "iguais"! e isso faz parte também da probabilidade.

```python
# Começando pelas bibliotecas
# Precisamos da biblioteca do Numpy para fazer calculos com baixo custo operacional, haja visto que roda C por baixo disso.
import numpy as np

# Importamos a bibliotexa do scipy.stats e importamos em especifico o ttest_ind para fazer a análise das amostras
from scipy.stats import ttest_ind

num_simulations = 10000

sample_size = 30

true_mean = 50

true_std = 10

significance_level = 0.01

significant_results = 0


for _ in range(num_simulations):

    group1 = np.random.normal(true_mean, true_std, sample_size)
    group2 = np.random.normal(true_mean, true_std, sample_size)

    t_stat, p_value = ttest_ind(group1, group2)

    if p_value < significance_level:
        significant_results += 1

proportion_significant = significant_results / num_simulations

# Quantidade de grupos testados entre si
print(f"Number of simulations: {num_simulations}")

# Nível de confiança estabelecido
print(f"Significance level (alpha): {significance_level}")

print(f"Proportion of significant results (Type I error rate): {proportion_significant:.4f}")
```

```python
#Importando a biblioteca pra calcular o t critico
import scipy.stats

# Primeiro definindo a amostra
batimentos = df['HeartRate']

# Peguei uma amostra não ordenada para análisar
amostra_n = batimentos[:50]

# Média da amostra
media_amostra = np.mean(amostra_n)

# Desvio padrão da amostra
s = np.std(amostra_n)

# Tamanho da amostra
n = len(amostra_n)

print("Média da amostra: ", media_amostra)
print("Desvio padrão da amostra: ", s)
print("Tamanho da amostra: ", n)

# Confiança da amostra
confianca = 0.95
alfa = 1 - confianca

# O t* é calculado baseado na tabela, considerando que tem os limites inferior e superio vamos dividir por 2
# A função ppf encontra o valor crítico na tabela T automaticamente (Isso aqui tive que ver aqui: https://www.geeksforgeeks.org/python/how-to-find-the-t-critical-value-in-python/)
t_critico = st.t.ppf(1 - alfa/2, df=n-1)

# Agora só montar o intervalo calculando os extremos da formula
limite_inferior = media_amostra - (t_critico * (s / np.sqrt(n)))
limite_superior = media_amostra + (t_critico * (s / np.sqrt(n)))

print("Limite Inferior: ", limite_inferior.round(3))
print("Limite Superior: ", limite_superior.round(3))
```
